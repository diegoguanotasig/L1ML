```python
# Eliminar la columna 'deck'
titanic = titanic.drop('deck', axis=1)

# Rellenar los valores nulos en 'age' con el promedio
titanic['age'].fillna(titanic['age'].mean(), inplace=True)

print("Columna 'deck' eliminada y valores nulos de 'age' rellenados.")
# Verificar los cambios mostrando la información del DataFrame nuevamente o los primeros valores.
titanic.info()
```

```python
import plotly.graph_objects as go

# Seleccionar columnas relevantes para el análisis de correlación
# Incluimos 'survived' como la variable objetivo
relevant_cols = ['survived', 'pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'class']
titanic_corr = titanic[relevant_cols].copy()

# Codificar la columna 'sex' (Female: 0, Male: 1) o (Male: 0, Female: 1)
titanic_corr['sex'] = titanic_corr['sex'].map({'female': 1, 'male': 0})

# Codificar la columna 'class' (Ordinal: First: 1, Second: 2, Third: 3)
titanic_corr['class'] = titanic_corr['class'].map({'First': 1, 'Second': 2, 'Third': 3})

# Calcular la matriz de correlación
correlation_matrix = titanic_corr.corr()

# Crear el heatmap de correlación con Plotly
fig = go.Figure(data=go.Heatmap(
                   z=correlation_matrix.values,
                   x=correlation_matrix.columns,
                   y=correlation_matrix.columns,
                   colorscale='Viridis',
                   colorbar=dict(title='Correlación')))

fig.update_layout(
    title='Matriz de Correlación del Dataset Titanic (Variables Clave)',
    xaxis_title='Variables',
    yaxis_title='Variables',
    xaxis_nticks=len(correlation_matrix.columns),
    yaxis_nticks=len(correlation_matrix.columns)
)

fig.show()
```
